```python
import nltk
from collections import Counter
import pandas as pd
import spacy

# Download NLTK data for tokenization
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Read the input text file
file_path = '/content/4.1_4.2_input.txt'
with open(file_path, 'r', encoding='utf-8') as file:
    text = file.read()

print("Input text loaded successfully.")
```

### Term-Frequency Analysis

```python
# Tokenize the text
words = nltk.word_tokenize(text.lower())

# Remove non-alphabetic tokens and common stop words (optional, but good for TF analysis)
# For simplicity, let's keep all words for now as the prompt doesn't specify stop words.
filtered_words = [word for word in words if word.isalpha()]

# Calculate word frequencies
word_counts = Counter(filtered_words)

print("Term frequencies calculated.")
```

```python
# Save the output in CSV format (Term, Frequency)
tf_df = pd.DataFrame(word_counts.items(), columns=['Term', 'Frequency'])
tf_csv_path = 'term_frequencies.csv'
tf_df.to_csv(tf_csv_path, index=False)

print(f"Term frequencies saved to {tf_csv_path}")
```

```python
# Identify the top 10 most frequent terms and display their corresponding frequencies
top_10_terms = tf_df.sort_values(by='Frequency', ascending=False).head(10)

print("\nTop 10 most frequent terms:")
display(top_10_terms)
```

### Named Entity Recognition (NER)

```python
# Load spaCy English model
nlp = spacy.load('en_core_web_sm')

# Process the text with spaCy
doc = nlp(text)

# Extract named entities
entities = []
for ent in doc.ents:
    entities.append((ent.text, ent.label_))

ner_df = pd.DataFrame(entities, columns=['Entity', 'Type'])

print("Named Entities identified.")
display(ner_df)
```

## 4.2 Term-Frequency Analysis (without any toolkit)

```python
import pandas as pd
import string

# Read the input text file again (ensure it's clean for this task)
file_path = '/content/4.1_4.2_input.txt'
with open(file_path, 'r', encoding='utf-8') as file:
    text_no_toolkit = file.read()

print("Input text re-loaded for toolkit-free analysis.")
```

```python
# 1. Convert text to lowercase
cleaned_text = text_no_toolkit.lower()

# 2. Remove punctuation and tokenize manually
# Create a translation table to remove punctuation
translation_table = str.maketrans('', '', string.punctuation)
cleaned_text = cleaned_text.translate(translation_table)

# Tokenize by splitting on whitespace
words_no_toolkit = cleaned_text.split()

# Filter out empty strings if any result from multiple spaces
words_no_toolkit = [word for word in words_no_toolkit if word]

print(f"Text cleaned and tokenized into {len(words_no_toolkit)} words.")
```

```python
# Calculate word frequencies manually
word_counts_no_toolkit = {}
for word in words_no_toolkit:
    word_counts_no_toolkit[word] = word_counts_no_toolkit.get(word, 0) + 1

print("Term frequencies calculated without toolkit.")
```

```python
# Save the output in CSV format (Term, Frequency)
tf_df_no_toolkit = pd.DataFrame(word_counts_no_toolkit.items(), columns=['Term', 'Frequency'])
tf_csv_path_no_toolkit = 'term_frequencies_no_toolkit.csv'
tf_df_no_toolkit.to_csv(tf_csv_path_no_toolkit, index=False)

print(f"Term frequencies (no toolkit) saved to {tf_csv_path_no_toolkit}")
```

```python
# Identify the top 10 most frequent terms and display their corresponding frequencies
top_10_terms_no_toolkit = tf_df_no_toolkit.sort_values(by='Frequency', ascending=False).head(10)

print("\nTop 10 most frequent terms (without toolkit):")
display(top_10_terms_no_toolkit)
```

## 4.3 TF-IDF Calculation (without NLTK, spaCy, or any other NLP toolkit)

```python
import math
import string

# Define the documents as specified in the prompt
documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}

processed_documents = {}
all_terms_in_corpus = set()

print("--- Preprocessing Documents ---")
for doc_name, text_content in documents.items():
    print(f"Processing {doc_name}:")

    # Convert text to lowercase
    text_lower = text_content.lower()
    print(f"  Lowercase: {text_lower}")

    # Remove punctuation and tokenize
    translation_table = str.maketrans('', '', string.punctuation)
    text_no_punc = text_lower.translate(translation_table)
    print(f"  No Punctuation: {text_no_punc}")

    # Tokenize the documents
    tokens = text_no_punc.split()
    # Filter out empty tokens if any
    tokens = [token for token in tokens if token]
    print(f"  Tokens: {tokens}")

    processed_documents[doc_name] = tokens
    all_terms_in_corpus.update(tokens)
print("\nAll documents preprocessed successfully.")
```

```python
print("\n--- Calculating Term Frequency (TF) ---")
tf_scores = {}
for doc_name, tokens in processed_documents.items():
    term_counts = {}
    for term in tokens:
        term_counts[term] = term_counts.get(term, 0) + 1

    total_terms_in_doc = len(tokens)
    tf_doc = {}
    for term, count in term_counts.items():
        tf_doc[term] = count / total_terms_in_doc
    tf_scores[doc_name] = tf_doc

    print(f"\n{doc_name} TF scores:")
    for term, score in tf_doc.items():
        print(f"  Term: '{term}', TF: {score:.4f}")
```

```python
print("\n--- Calculating Document Frequency (DF) ---")
df_scores = {}
for term in all_terms_in_corpus:
    df_scores[term] = 0
    for doc_name, tokens in processed_documents.items():
        if term in tokens:
            df_scores[term] += 1

print("DF scores:")
for term, score in df_scores.items():
    print(f"  Term: '{term}', DF: {score}")
```

```python
print("\n--- Calculating Inverse Document Frequency (IDF) ---")
num_documents = len(documents)
idf_scores = {}
for term, df in df_scores.items():
    idf_scores[term] = math.log10(num_documents / df)

print("IDF scores:")
for term, score in idf_scores.items():
    print(f"  Term: '{term}', IDF: {score:.4f}")
```

```python
print("\n--- Calculating TF-IDF for every term in every document ---")
tfidf_scores = {}
for doc_name, tf_doc in tf_scores.items():
    tfidf_doc = {}
    for term, tf in tf_doc.items():
        idf = idf_scores[term]
        tfidf_doc[term] = tf * idf
    tfidf_scores[doc_name] = tfidf_doc

    print(f"\n{doc_name} TF-IDF scores:")
    for term, tfidf in tfidf_doc.items():
        tf = tf_scores[doc_name][term]
        df = df_scores[term]
        idf = idf_scores[term]
        print(f"  Term: '{term}', TF: {tf:.4f}, DF: {df}, IDF: {idf:.4f}, TF-IDF: {tfidf:.4f}")
```

```python
print("\n--- Identifying Top 10 terms with highest TF-IDF score for each document ---")
for doc_name, tfidf_doc in tfidf_scores.items():
    # Sort terms by TF-IDF score in descending order
    sorted_terms = sorted(tfidf_doc.items(), key=lambda item: item[1], reverse=True)
    top_10_terms = sorted_terms[:10]

    print(f"\nTop 10 TF-IDF terms for {doc_name}:")
    for term, score in top_10_terms:
        print(f"  Term: '{term}', TF-IDF: {score:.4f}")
```
