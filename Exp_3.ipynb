## Experiment 3.1: Stemming using Porter Stemmer

```python
import nltk
from nltk.stem import PorterStemmer

nltk.download('punkt')

stemmer = PorterStemmer()

words_for_stemming = [
    'playing', 'played', 'plays', 'studies', 'studying',
    'connected', 'connection', 'computers'
]

print('Original Word | Stemmed Word')
print('--------------------------')
for word in words_for_stemming:
    stemmed_word = stemmer.stem(word)
    print(f'{word:<13} | {stemmed_word}')
```

## Experiment 3.2: Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words_for_lemmatization = [
    'cats', 'dogs', 'running', 'runs', 'ran', 'studies',
    'studying', 'better', 'children', 'mice', 'went', 'ate',
    'leaves', 'caring'
]

print('Original Word | Lemmatized Word (noun) | Lemmatized Word (verb) | Lemmatized Word (adj)')
print('--------------------------------------------------------------------------------------')
for word in words_for_lemmatization:
    lemma_n = lemmatizer.lemmatize(word, pos='n') # Noun
    lemma_v = lemmatizer.lemmatize(word, pos='v') # Verb
    lemma_a = lemmatizer.lemmatize(word, pos='a') # Adjective
    print(f'{word:<13} | {lemma_n:<20} | {lemma_v:<22} | {lemma_a}')
```

## Experiment 3.3: Regular Expressions (Regex) for Information Extraction

```python
import re

input_text = """
Welcome to the Natural Language Processing workshop! For more information, visit https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your experience using #NLP, #Python, and #MachineLearning. You can also visit https://github.com/NLPWorkshop for the latest updates
"""

print("Input Text:")
print(input_text)
print("\n" + "="*50 + "\n")

# 1. Extract Email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, input_text)
print(f"Email Addresses: {emails}")

# 2. Extract URLs
url_pattern = r'\b(?:https?://|www\.)\S+\b'
urls = re.findall(url_pattern, input_text)
print(f"URLs: {urls}")

# 3. Extract Mobile numbers (simple pattern for +91-XXXXXXXXXX or XXXXXXXXXX)
mobile_pattern = r'(?:\+\d{1,3}-)?\d{10}\b'
mobile_numbers = re.findall(mobile_pattern, input_text)
print(f"Mobile Numbers: {mobile_numbers}")

# 4. Extract Hashtags
hashtag_pattern = r'#\w+'
hashtags = re.findall(hashtag_pattern, input_text)
print(f"Hashtags: {hashtags}")

# 5. Extract Mentions
mention_pattern = r'@\w+'
mentions = re.findall(mention_pattern, input_text)
print(f"Mentions: {mentions}")
```
