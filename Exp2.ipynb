## Experiment 2.1: Basic Text Preprocessing

```python
import re
import string

# Load the text from the file
file_path_2_1 = '/content/2.1_text_data.txt'
with open(file_path_2_1, 'r') as f:
    text_2_1 = f.read()

print("Original Text:")
print(text_2_1)
print("\n" + "="*50 + "\n")

# 1. Convert text to lowercase. (Also display how many uppercase?)
initial_uppercase_count = sum(1 for char in text_2_1 if char.isupper())
text_lowercase = text_2_1.lower()
print(f"Number of uppercase characters: {initial_uppercase_count}")

# 2. Remove punctuation. (Also display how many different types of punctuations?)
punctuation_chars = set()
for char in text_2_1:
    if char in string.punctuation:
        punctuation_chars.add(char)

text_no_punctuation = ''.join([char for char in text_lowercase if char not in string.punctuation])
print(f"Different types of punctuations found: {', '.join(sorted(list(punctuation_chars)))}")
print(f"Number of different punctuation types: {len(punctuation_chars)}")

# 3. Remove numbers. (Also display how many numbers?)
number_count = sum(1 for char in text_2_1 if char.isdigit())
text_no_numbers = re.sub(r'\d+', '', text_no_punctuation)
print(f"Number of numbers found: {number_count}")

# 4. Remove extra whitespace. (How much extra whitespace is present?)
# We'll count extra whitespaces by looking for sequences of two or more spaces or newlines/tabs
extra_whitespace_count = len(re.findall(r'\s{2,}', text_2_1))
text_cleaned = re.sub(r'\s+', ' ', text_no_numbers).strip()
print(f"Number of extra whitespace sequences (2 or more spaces/newlines/tabs): {extra_whitespace_count}")

# 5. Display the cleaned text.
print("\nCleaned Text:")
print(text_cleaned)
```

## Experiment 2.2: Word and Sentence Tokenization

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
import spacy

# Download NLTK data (if not already downloaded)
try:
    nltk.data.find('tokenizers/punkt_tab') # Check for punkt_tab specifically
except LookupError:
    nltk.download('punkt_tab')

# Load spaCy model
nlp = spacy.load('en_core_web_sm')

# Load the text from the file
file_path_2_2 = '/content/2.2_tokenization_data.txt'
with open(file_path_2_2, 'r') as f:
    text_2_2 = f.read()

print("Original Text for Tokenization:")
print(text_2_2)
print("\n" + "="*50 + "\n")

# --- 1. Without using NLTK/spaCy ---
print("### Tokenization without NLTK/spaCy ###")
# Simple word tokenization: split by whitespace and remove empty strings
words_manual = [word for word in text_2_2.split() if word.strip()]
print("Manual Word Tokens:", words_manual)

# Simple sentence tokenization: split by common sentence delimiters (. ! ?) followed by whitespace
sentences_manual = re.split(r'(?<=[.!?])\s+', text_2_2)
# Clean up any empty strings from splitting
sentences_manual = [s.strip() for s in sentences_manual if s.strip()]
print("Manual Sentence Tokens:", sentences_manual)
print("\n" + "-"*30 + "\n")

# --- 2. Using NLTK ---
print("### Tokenization using NLTK ###")
nltk_word_tokens = word_tokenize(text_2_2)
print("NLTK Word Tokens:", nltk_word_tokens)

nltk_sent_tokens = sent_tokenize(text_2_2)
print("NLTK Sentence Tokens:", nltk_sent_tokens)
print("\n" + "-"*30 + "\n")

# --- 3. Using spaCy ---
print("### Tokenization using spaCy ###")
doc = nlp(text_2_2)
spacy_word_tokens = [token.text for token in doc]
print("spaCy Word Tokens:", spacy_word_tokens)

spacy_sent_tokens = [sent.text for sent in doc.sents]
print("spaCy Sentence Tokens:", spacy_sent_tokens)
```

## Experiment 2.3: Tokenization and Stop Word Removal

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download NLTK stopwords (if not already downloaded)
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

# Download NLTK punkt_tab for word_tokenize (if not already downloaded)
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

# Load the text from the file
file_path_2_3 = '/content/2.3_clean_data.txt'
with open(file_path_2_3, 'r') as f:
    text_2_3 = f.read()

print("Original Text for Stop Word Removal:")
print(text_2_3)
print("\n" + "="*50 + "\n")

# Tokenize the text
original_tokens = word_tokenize(text_2_3.lower()) # Convert to lowercase for consistent stop word matching
print("Original Tokens:")
print(original_tokens)
print("\n" + "-"*30 + "\n")

# Get English stop words
stop_words = set(stopwords.words('english'))

# Filter out stop words
filtered_tokens = [word for word in original_tokens if word not in stop_words and word.isalpha()]

print("Filtered Tokens (after removing stop words and non-alphabetic tokens):")
print(filtered_tokens)
print("\n" + "-"*30 + "\n")

print("Stop Words Used:")
print(sorted(list(stop_words)))
```
