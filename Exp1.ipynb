# Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/employee_information_100.csv')

sns.set(style="whitegrid")
display(df.head())
```

### Experiment-1.1: Average salary of each department (Bar Chart)

```python
plt.figure(figsize=(10, 6))
avg_salary = df.groupby('Department')['Salary'].mean().sort_values()
avg_salary.plot(kind='barh', color='skyblue')
plt.title('Average Salary by Department')
plt.xlabel('Average Salary')
plt.ylabel('Department')
plt.show()
```

### Experiment 1.2: Number of employees in each department

```python
dept_counts = df['Department'].value_counts()
display(dept_counts)
```

### Experiment 1.3: Percentage of male and female employees (Pie Chart)

```python
plt.figure(figsize=(7, 7))
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['lightpink', 'lightblue'])
plt.title('Gender Distribution')
plt.ylabel('')
plt.show()
```

### Experiment 1.4: Salary distribution (Histogram)

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Salary'], kde=True, color='green')
plt.title('Salary Distribution')
plt.show()
```

### Experiment 1.5: Experience vs Salary (Scatter Plot)

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Experience_Years', y='Salary', data=df, hue='Gender')
plt.title('Relationship between Experience and Salary')
plt.show()
```

### Experiment 1.6: Top 10 highest-paid employees

```python
top_10 = df.nlargest(10, 'Salary')
display(top_10)
```

### Experiment 1.7: Highest salary in every department

```python
max_salary_dept = df.groupby('Department')['Salary'].max()
display(max_salary_dept)
```

### Experiment 1.8: Employees with salary > overall average

```python
avg_total = df['Salary'].mean()
above_avg = df[df['Salary'] > avg_total]
print(f"Overall Average: {avg_total}")
display(above_avg)
```

### Experiment 1.9: Average years of experience in each department

```python
avg_exp = df.groupby('Department')['Experience_Years'].mean()
display(avg_exp)
```

### Experiment 1.10: Age distribution (Histogram)

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=15, kde=True, color='orange')
plt.title('Employee Age Distribution')
plt.show()
```
